```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
```

# Dataset
[1] R. Bock. "MAGIC Gamma Telescope," UCI Machine Learning Repository, 2004. [Online]. Available: https://doi.org/10.24432/C52C8B.

# Personal Testing

```python
df = pd.read_csv('/content/gdrive/MyDrive/ML/MAGIC_Gamma/magic04.data', names=['fLength', 'fWidth', 'fSize', 'fConc', 'fConc1', 'fAsym', 'fM3Long', 'fM3Trans', 'fAlpha', 'fDist', 'class'])
X = df.drop('class', axis=1) # [1]
y = df['class']
y.replace('h', 0, inplace=True)
y.replace('g', 1, inplace=True)

# Create train and validation sets
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, shuffle=True)
```

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression

pipe = make_pipeline(StandardScaler(), LogisticRegression())
pipe.fit(X_train, y_train)
pipe.score(X_val, y_val)
```

```python
# Logistic Regression without StandardScalar() does slightly worse than regular logistic regression
# Struggles on large datasets with multidimensionality in features

clf = LogisticRegression()
clf.fit(X_train, y_train)
clf.score(X_val, y_val)
```

```python
# MLP (scikit) no standardization
from sklearn.neural_network import MLPClassifier

clf = MLPClassifier(max_iter=450).fit(X_train, y_train)
clf.predict(X_val)
clf.score(X_val, y_val)
```
